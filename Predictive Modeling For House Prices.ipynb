```python
x = data[['number of bedrooms', 'number of bathrooms', 'living area', 'lot area', 'number of floors','condition of the house','Area of the house(excluding basement)', 'Area of the basement', 'Number of schools nearby', 'Distance from the airport',]]
y = data[["Price"]]
```

```python
data.shape[0]  * 0.8
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2)
```

```python
x_train.shape
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.tree import DecisionTreeRegressor
parma_grid = {
    "criterion" : ["mse", "friedman_mse", "mae"],
    "splitter" : ["best", "random"],
    "max_depth" : [None,10,20,30,40,50],
    "min_samples_split" : [2,5,10],
    "min_samples_leaf" : [1,2,4]
    }
```

```python
tree_model = DecisionTreeRegressor()
```

```python
grid_tree = GridSearchCV(estimator = tree_model, param_grid= parma_grid)
```

```python
grid_tree.fit(x_train, y_train)
```

```python
grid_tree.best_params_
```

```python
tree_preds = grid_tree.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test, tree_preds)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

```python
lr.fit (x_train, y_train)
```

```python
predslr = lr.predict(x_test)
```

```python
mean_squared_error(y_test, predslr)
```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test, tree_preds)
```

```python
mean_absolute_error(y_test, predslr)
```

```python
from sklearn.ensemble import RandomForestRegressor
rfmodel = RandomForestRegressor()
```

```python
param_gridrfr = {
    "max_depth" : (5,10,15),
    "n_estimators" : (2,3,4,5,6,7,8,9,10),
}
```

```python
gridrfr = GridSearchCV(rfmodel, param_gridrfr)
```

```python
gridrfr.fit(x_train, y_train.values.ravel())
```

```python
gridrfr.best_params_
```

```python
rfpredictions = gridrfr.predict(x_test)
```

```python
mean_absolute_error(y_test, rfpredictions)
```
